## Data Cleaning
-----
### Tasks:
1. Identify and remove Duplicates
2. Locate Data with Missing values
3. Data type conversion, convert Data Types where needed.

````python
# Remove un-need tabled
# sol.pop('omega') # Slow method

# Creating List to check if columns exist
to_keep = ['name', 'planet_status', 'mass', 'radius', 'discovered','updated','star_name','star_age','star_sp_type','star_distance']
def chk_df_col(x, to_keep):
    if not set(to_keep).issubset(set(x.columns)):
        return f"{' and '.join(set(to_keep).difference(x.columns))} are not available in the dataframe"
    return "Avalible"

chk_df_col(df_sol, to_keep) # Runs to check if all columns are there
````

````python
# Removing columns to saved in the 'to_keep' list
df_sol = sol[sol.columns.intersection(to_keep)]
df_sol.info()
````

````python
# Duplicates
df_sol.duplicated(['name']).info()
````

````python
# Checking planet_status as a "Confirmed" Exoplanet
df_sol[sol.planet_status.str.contains("Candidate")]
````

````python
# Missing values
````

````python
# Removing ExoPlanets with missing data regarding star type, star age, distance to star, mass
df_sol.dropna(subset=["star_sp_type", "star_age", "star_distance", "mass"], inplace = True)
df_sol
````

````python
# Data type conversion
````

````python
# Renaming tables
````

````python
# Changing sp_star_type -> star_type and keeping the classification of star to first letter and not first letter sub type 
df_sol = df_sol.rename(columns={"star_sp_type": "star_type"})
df_sol
````

## Data analysis
-----
### Task:
1. Develop at least 5 analytical questions
2. Use Pandas to answer those questions

````python
# Question 1: What planets are likely Earth replacements
# Converting mass from jupiter to earth mass kg (https://www.vcalc.com/wiki/vCalc/Mass-of-the-Earth)
df_e_mass = df_sol["mass"] /  317.827601
df_e_mass # e_mass = Earth mass. 317.827601 is the rough differences between Earth and Jupiter masses
````

````python
# Question 2: Solar systems within 153.3007 Parsecs (500 Light years) and the ages of the stars.
df_plnt_dist = df_sol[df_sol['star_distance'].between(0, 153.3007)]
df_plnt_dist # DataFrame to display distance in Parsecs for travel up to 500 light years
````

````python
# Question 3: Names of the closest planet to ours, within a set age of star
Q3_Filter = (
    df_sol[(df_sol["star_age"] >= 0) & (df_sol["star_age"] <= 20)]
    .sort_values("star_distance")
    .head(20)
)

Q3_Filter
````

````python
# Question 4: Stars with the most exoplanets
exo_counts = df_sol["star_name"].value_counts()

top_stars = exo_counts.head(10).index

most_exo = df_sol[df_sol["star_name"].isin(top_stars)]
most_exo["exo_count"] = most_exo["star_name"].map(exo_counts)
most_exo = most_exo.sort_values("exo_count", ascending=True)
most_exo
````

````python
# Question 5: Most common Star type
df_sol["star_type"] = df_sol["star_type"].str[0]
most_common = df_sol["star_type"].value_counts().idxmax()
f"Most common star type letter: {most_common}"
df_sol["star_type"].value_counts()
````

## Visualistions

````python
# Question 1:
fig, ax = plt.subplots(figsize=(10, 7))

sns.histplot(
    data=df_sol,
    x="mass",
    y="star_distance",
    bins=30,
    cmap="YlOrRd",
    ax=ax
)

ax.set_xscale("log")
ax.set_xlabel("Mass (Jupiter masses, log)")
ax.set_ylabel("Star Distance (ly)")
ax.legend()
ax.set_title("Planet Mass vs Star Distance")
````

````python
# Question 2: Solar systems within 153.3007 Parsecs (500 Light years) and the ages of the stars.
Q_Two = sns.relplot(data=df_plnt_dist, x="star_age", y="star_distance", color="m")
Q_Two.fig.suptitle("Distance of System, and age of their star")
Q_Two.fig.subplots_adjust(top=0.9)
Q_Two.set_axis_labels("Distance of System", "Age of Star")
````

````python
# Question 3: Names of the closest planet to ours, within a set age of star
sns.barplot(data=Q3_Filter, x="name", y="star_distance", color="m")
plt.title("Exoplanets near Earth with stars up to 20million years")
plt.xlabel("Planet")
plt.ylabel("Distance (light-years)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
````

````python
# Question 4: Stars with the most exoplanets
sns.lineplot(x="star_name", y="exo_count", data=most_exo, marker="o")
plt.xticks(rotation=45, ha="right")
plt.title("Exoplanet Count per Star")
````

````python
# Question 5: Most common Star type
counts = df_sol["star_type"].value_counts()[lambda x: x > 20]

fig, ax = plt.subplots(figsize=(8, 6))

ax.pie(counts, labels=counts.index, startangle=90, colors=sns.color_palette("tab10"))
ax.set_title("Most Common Star Types")
````
